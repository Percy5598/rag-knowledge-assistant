# Embeddings: Turning Meaning into Numbers
1. What is embedding?
An embedding is a numerical vector representing some aspect of the meaning or linguistic properties of text.

2. Why embeddings are different from the TF-IDF?
TF-IDF answers "Which words are important in this document?" but embeddings encodes relationships learned from language data.

```python
import sys
print(sys.executable)
```

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer("all-MiniLM-L6-v2") 
sentences = [
    "Employees receive 30 days of annual leave.",
    "Workers get 30 vacation days every year.",
    "The company sells software products.",
    "The weather is sunny today."
]
embeddings = model.encode(sentences)
embeddings.shape
```

```python
print(embeddings[0])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similarity = cosine_similarity(embeddings)
similarity 
```

RAG = retrieve information at inference time, then generate using that information. LLM does not learn from given documents.

Why can't TF-IDF reliably identify that "annual leave" and "vacation days" are related?
TF-IDF uses the frequency as numerical representation based on frequency of the word to show the importance of the word in the documents.

What does an embedding give us that TF-IDF doesn't?
Embedding uses  similarity matrix as numerical representation to show the semantic similarity beetween the words.

Sparse representation: Usually many zeros i.e., [0, 0, 0, 0, 0, 0, 0, 1, 0, ...]
dense representation: Many dimension non zeros i.e., [-0.034, 0.081, 0.017, -0.092, ...]
