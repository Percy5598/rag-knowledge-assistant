## Tokenization 
Breaking text into meaningful pieces. LLMs don't receive unlimited raw human text. They operate in tokens. You cannot give all the tokens to an LLM everytime someone query about something. So RAG system i.e., retrieval comes into play.      

10,000 pages
      ↓
retrieve relevant pieces
      ↓
perhaps 3–5 chunks
      ↓
LLM

```python
tokens = text.split()
tokens[:20]
```

```python
tokens = text.split()

print("Number of tokens:", len(tokens))
print("First 10 tokens:", tokens[:10])
```

1. What is tokens?
A unit of text that a tokenizer splits a piece of text into for processing.

2. Why isn't text.split() a perfect tokenizer?
Whitespace-separated words are not necessarily the same thing as NLP tokens.

3. Why can't we simply give the entire corporate document repository to the LLM?
