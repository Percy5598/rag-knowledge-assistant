## Vocabulary 
A vocabulary is the set of unique tokens observed in a collection of text.

```python
documents = [
    "employees receive annual leave",
    "employees request unpaid leave",
    "managers approve annual leave"
]
tokenized_documents = []

for document in documents:
    tokens = document.split()
    tokenized_documents.append(tokens)

tokenized_documents
```

```python
vocabulary = set()
for tokens in tokenized_documents:
    for token in tokens:
        vocabulary.add(token)

vocabulary
```

```python
vocabulary = sorted(vocabulary)
vocabulary
```

## Bag of Words 
Represent a document by counting how often vocabulary words occur.

```python
# This tells does the word exit in the tokens 
document = "employees receive annual leave"

tokens = document.split()

vector = []

for word in vocabulary:
    if word in tokens:
        vector.append(1)
    else:
        vector.append(0)

vector
```

```python
# THis counts the words in the document that are present in the token document.
document = "leave leave leave annual"

tokens = document.split()

vector = []

for word in vocabulary:
    count = tokens.count(word)
    vector.append(count)

vector
```
