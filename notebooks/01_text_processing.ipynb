## Vocabulary 
A vocabulary is the set of unique tokens observed in a collection of text.

```python
documents = [
    "employees receive annual leave",
    "employees request unpaid leave",
    "managers approve annual leave"
]
tokenized_documents = []

for document in documents:
    tokens = document.split()
    tokenized_documents.append(tokens)

tokenized_documents
```

```python
vocabulary = set()
for tokens in tokenized_documents:
    for token in tokens:
        vocabulary.add(token)

vocabulary
```

```python
vocabulary = sorted(vocabulary)
vocabulary
```

## Bag of Words 
Represent a document by counting how often vocabulary words occur.

```python
# This tells does the word exit in the tokens 
document = "employees receive annual leave"

tokens = document.split()

vector = []

for word in vocabulary:
    if word in tokens:
        vector.append(1)
    else:
        vector.append(0)

vector
```

```python
# THis counts the words in the document that are present in the token document.
document = "leave leave leave annual"

tokens = document.split()

vector = []

for word in vocabulary:
    count = tokens.count(word)
    vector.append(count)

vector
```

## TF-IDF: Finding Important Words
It stands for Term FRequency - Inverse Document Frequency
Term Frequency answers "How often the word appears in this document?"
Inverse Document Frequency explains " How rare is this word across the document collection?"

A common simplified formulation is: 
TF(t, d) = (count of t in d) / (total terms in d) 
IDF(t)   = log(N / df(t))

where,

N = number of documents
df = number of documents containing term t

Then,

TFIDF(t, d) =  TF(t,d) * IDF(t)

```python
documents = [
    "employees receive annual leave",
    "employees request unpaid leave",
    "managers approve annual leave"
]

tokenized_documents = [
    document.split()
    for document in documents
]
tokenized_documents
```

```python
N = len(documents)
print("Number of documents:", N)
```

```python
vocabulary = sorted(
    set(
        token
        for tokens in tokenized_documents
        for token in tokens
    )
)
vocabulary
```

```python
document_frequency = {}

for word in vocabulary:
    count = 0

    for tokens in tokenized_documents:
        if word in tokens:
            count += 1

    document_frequency[word] = count

document_frequency
```

```python
import math
idf = {}

for word, df in document_frequency.items():
    idf[word] = math.log(N / df)

idf
```

```python
# Cosine Similarity: Mathematical way to compare the vectors

def cosine_similarity(vector_a, vector_b):
    n = sum(a * b for a, b in zip(vector_a, vector_b))
    d = math.sqrt(sum(x * x for x in vector_a))
    result = n/d
    return result
```
